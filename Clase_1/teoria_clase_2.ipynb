# Condicionales y bucles

>Condicionales → if/elif/else, es una estructura de control del flujo del programa
si yo le envio el taller hoy entonces usted no va a dormir
si yo se lo envio mañana usted duerme hoy 
palabras q use: si, entonces, sino
if es si no es como preguntonsito

```python
nombre = "Ana"
promedio = 8.7

if promedio >= 7: #el if lo q evalua realmente es si es un booleano si es true o false
    print(f"felicitaciones {nombre}")
#ahora todo lo q no entre en la primera condicion entra aqui
else: #sino
    print(f"necesitas mejorar")
```

```python
nombre = "Ana"
promedio = 5

if promedio >= 7: 
    print(f"felicitaciones {nombre}")
elif promedio >=5:  
    print(f"{nombre} Puedes mejorar")
else: 
    print(f"necesitas mejorar")
```

## Bucles

>For, es una estructura de iteracion es decir recorre todo el elemento las veces q yo le indique


```python
#hacer una lista y dentro va a contener diccionarios - esta es la mejor manera a diferncia de las de arriba 
learners = [
    {"nombre": "Pepita", "Promedio": 8.7},
    {"nombre": "carlos", "Promedio": 7.5},
    {"nombre": "william", "Promedio": 9},
]

for learner in learners: 
#para un estudiante que esta en la lista estudiantes - for itera sobre cada uno de los elmentos de la lista
#entonces cuando el for itera por primera vez learner va a llamarse pepita 
    if learner["Promedio"] >= 7:  # learner quiere decir si la persona q tomo en ese momento la posicion tiene en su promedio
        print(f"felicitaciones {learner ['nombre']}")
    elif learner ["Promedio"] >=5:  
        print(f"{learner ['nombre']} Puedes mejorar")
    else: 
        print(f"necesitas mejorar")



```
